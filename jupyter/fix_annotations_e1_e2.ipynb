This notebook was used to fix ANNOTATIONS (not images, which were lost...) of a small dataset. That dataset (named E1 and E2) was created in one of the first concepts of experiment setup which assumed that every fasade has its own distinct ratio parameter, which could extend to red AND green channels (value up to 5.10), and every value meant different class.

```python
import json
import os
```

```python
ANNOTATIONS_DIR = "C:\\Users\\WA\\Desktop\\misc\\annotation_E1_E2_old"
ANNOTATIONS_DIR_OUT = "C:\\Users\\WA\\Desktop\\misc\\annotation_E1_E2_new"


annotations_files = list(os.listdir(ANNOTATIONS_DIR))
```

```python
def get_true_value(initial_value):
    return int(initial_value*0.825)


def fix_annotations(annotations: dict, a_type:str) -> dict:
    fixed_annotations = annotations.copy()
    fixed_annotations["annotations"] = []
    for annotation_object in annotations["annotations"]:
        true_value = get_true_value(annotation_object["parameter_value"])
        annotation_object["parameter_value"] = true_value
        if annotation_object["category_id"] >= 0:             
            if a_type == "normal":
                annotation_object["category_id"] = true_value
            else:
                annotation_object["category_id"] = true_value // 25
        else:
            annotation_object["category_id"] = 1000
        fixed_annotations["annotations"].append(annotation_object)

    categories = [{
         "id": 1000,
         "name": "Facade",
         "supercategory": "object"
      }] 
    
    if a_type == "normal":
        for i in range(0, 511):
            categories.append({"id": i, "name": str(i), "supercategory": "object"})
    else:
        for i in range(0, 511//25):
            categories.append({"id": i, "name": str(i), "supercategory": "object"})
    
    fixed_annotations["categories"] = categories

    return fixed_annotations


for annotations in annotations_files:
    absolute_path = os.path.join(ANNOTATIONS_DIR, annotations) 
    #print("ANNS:", absolute_path)
    with open (absolute_path) as f:
        annotations_dict = json.load(f)
    if "smooth" in annotations:
        new_annotations = fix_annotations(annotations_dict, a_type="smooth")
    else:
        new_annotations = fix_annotations(annotations_dict, a_type="normal")

    absolute_out_path = os.path.join(ANNOTATIONS_DIR_OUT, annotations) 
    with open (absolute_out_path, mode="w") as f:
        json.dump(new_annotations, f, indent=2)

print("DONE" + "="*10)
```
