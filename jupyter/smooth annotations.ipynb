```python
import json
import os
from pathlib import Path

print(os.getcwd())
```

```python
ANNOTATION_PATH = Path("../dataset_coco/annotations")

def smooth_annotations(split, smooth_factor):
    annotations_m1 = ANNOTATION_PATH / f"instances_{split}_m1.json"
    annotations_m2 = ANNOTATION_PATH / f"instances_{split}_m2.json"

    annotations_m1_smooth = ANNOTATION_PATH / f"instances_{split}_smooth_m1.json"
    annotations_m2_smooth = ANNOTATION_PATH / f"instances_{split}_smooth_m2.json"

    #print(annotations_m1.exists(), annotations_m2.exists())
    for annotations_original, annotations_smooth in zip([annotations_m1, annotations_m2], [annotations_m1_smooth, annotations_m2_smooth]):
        with open (annotations_original, mode='r') as f:
            annotations = json.load(f)
        
       #print(len(annotations["annotations"]))
        annotations_copy = annotations.copy()
        annotations_copy["annotations"] = []
        annotations_copy["categories"] = []
        
        for annotation_object in annotations["annotations"]:
            if annotation_object["category_id"] >= 0:
                annotation_object["category_id"] = annotation_object["category_id"] // smooth_factor
            annotations_copy["annotations"].append(annotation_object)


        ids_present = []

        for annotation_object in annotations["categories"]:
            if annotation_object["id"] >= 0:
                annotation_object["id"] = annotation_object["id"] // smooth_factor
                annotation_object["name"] = str(annotation_object["id"])
            if annotation_object["id"] not in ids_present:
                annotations_copy["categories"].append(annotation_object)
                ids_present.append(annotation_object["id"])
        
        annotations_copy["categories"] = list(annotations_copy["categories"])

        with open(annotations_smooth, mode="w") as file:
            json.dump(annotations_copy, file, indent=4)

    return



splits = ["train", "val", "test"]

smooth_factor = 25
for split in splits:
    #print("SPLIT:", split)
    smooth_annotations(split, smooth_factor)

print("Done " + "="*10)
```
