```python
from pathlib import Path
import re

IMAGES_DIR =  Path("../images")
ALL_FILES = [path for path in os.listdir(IMAGES_DIR) if re.search(".*\.png", path) and len(path.split("_")) == 6]
COUNTER_MAP = {'0': 0}

def set_counter_mapping(city_index):
    n_of_cities = {city for city, _ in city_index}
    for city in n_of_cities:
        COUNTER_MAP[str(int(city)+1)] = max([index for _, index in city_index])
    
    print(n_of_cities)
    print(COUNTER_MAP)


def get_new_index(current_index, n_city):
    return str(int(current_index) + int(COUNTER_MAP[n_city]))

def rename_photo(path):
    path_split = path.split("_")
    new_path = path_split[0] + "_" + path_split[1] + "_" + path_split[3] + "_" + path_split[4] + "_" + get_new_index(path_split[5][:-4], path_split[2]) + ".png"
    print("OLD:", IMAGES_DIR / path)
    print("NEW:", IMAGES_DIR / new_path)

    try:
        os.rename(IMAGES_DIR / path, IMAGES_DIR / new_path)
    except FileNotFoundError:
        print("file not found")
    except FileExistsError:
        print("file already exists")
    

#images\SynCity3D_m-1_<0>_snapshot_back_<2>.png

#OLD: SynCity3D_m-1_1_snapshot_back_1.png
#NEW: SynCity3D_m-1_snapshot_back_11.png


#for file in ALL_FILES[:100]:
#    print(file)

print("1...")
city_index = [(path.split('_')[2], int(path.split('_')[5][:-4]) ) for path in ALL_FILES]
#city_index = [(path.split('_'), len(path.split('_'))) for path in ALL_FILES]

print("2...")
print(city_index)


print("3...")

set_counter_mapping(city_index)

print("4...")
#rename_photo(ALL_FILES[1])
for file in ALL_FILES:
    rename_photo(file)

```
