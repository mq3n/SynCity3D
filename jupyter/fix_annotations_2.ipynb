This notebook changes the annotation structure so annotations of M2 could be used for double semantic segmentation (segment=Facade/Window, parameter_value=2.11).

```python
import json 
import os
from pathlib import Path

print(os.getcwd())
```

```python
#ClassDef("Facade", 1000), ClassDef("Window", 1001)

ANNOTATIONS_DIRS = [Path("..\\COCO_DATASETS\\D1_2\\annotations\\DEFUALT"), 
                    Path("..\\COCO_DATASETS\\D1_2\\annotations\\SMOOTHED"), 
                    Path("..\\COCO_DATASETS\\D3\\annotations")]
```

M1 1

```python

for dir in ANNOTATIONS_DIRS[1:2]:
    m1_annotations = [dir / path for path in os.listdir(dir) if "m1" in path]
    m2_annotations = [dir / path for path in os.listdir(dir) if "m2" in path]

    #print("M1:")
    #print(m1_annotations)
    #print("M2")
    #print(m2_annotations)

    print("M1......................")

    for annotations in m1_annotations[:]:
        with open (annotations, mode="r") as f:
            ann = json.load(f)
        
        ann_fixed = ann.copy()
        ann_fixed["annotations"] = []
        ann_fixed["categories"] = []

        for ann_object in ann["annotations"]:
            
            if ann_object["category_id"] != 1000:
                ann_object["category_id"] = int(ann_object["category_id"]) + 1

            ann_fixed["annotations"].append(ann_object)


        for cat_object in ann["categories"]:
            if cat_object["id"] != 1000 and cat_object["id"] != 1001:

                bin_nr = cat_object["name"]
                cat_object["name"] = f"{1 + 25*int(bin_nr)}-{25 + 25*int(bin_nr)}"
                cat_object["id"] += 1


            if cat_object["id"] != 1001:
                ann_fixed["categories"].append(cat_object)
        
        #print("ANNOTATIONS:", annotations)
        #print("CATEGORIES:", ann_fixed["categories"])

        #print("ANNS:", )
        #for a in ann_fixed["annotations"][:3]:
        #    print(a)
        #    print()
        #print()

        with open (annotations, mode="w") as f:
            json.dump(ann_fixed, f, indent=2)



print("DONE", "="*10)
```

M2 1

```python
for dir in ANNOTATIONS_DIRS[1:2]:
    m1_annotations = [dir / path for path in os.listdir(dir) if "m1" in path]
    m2_annotations = [dir / path for path in os.listdir(dir) if "m2" in path]

    #print("M1:")
    #print(m1_annotations)
    #print("M2")
    #print(m2_annotations)

    for annotations in m2_annotations[:]:
        with open (annotations, mode="r") as f:
            ann = json.load(f)
        
        ann_fixed = ann.copy()
        ann_fixed["annotations"] = []
        ann_fixed["categories"] = []

        for ann_object in ann["annotations"]:
            

            segment_type, bin_value  = ann_object["category_id"] 
            if segment_type != 1000:
                ann_object["category_id"] = [segment_type, bin_value + 1]


            ann_fixed["annotations"].append(ann_object)

        for cat_object in ann["categories"]:
            if cat_object["id"] != 1000 and cat_object["id"] != 1001:
                #print("CAT OBJECT:", cat_object)
                #print()

                bin_nr = cat_object["name"]
                cat_object["name"] = f"{1 + 25*int(bin_nr)}-{25 + 25*int(bin_nr)}"
                cat_object["id"] += 1


            if cat_object["id"] != 1001:
                ann_fixed["categories"].append(cat_object)
        
        #print("ANNOTATIONS:", annotations)
        #print("CATEGORIES:", ann_fixed["categories"])

        #print("ANNS:", )
        #for a in ann_fixed["annotations"][:3]:
        #    print(a)
        #    print()
        #print()
    
        with open (annotations, mode="w") as f:
            json.dump(ann_fixed, f, indent=2)




print("DONE", "="*10)
```
